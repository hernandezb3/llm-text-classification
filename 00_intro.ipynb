# Classes
+ A class defines a new kind of value.

```python
from enum import Enum # defines a fixed set of allowed choices

class Answer(str, Enum): # create a new type of thing called an answer. because we are using Enum, that new thing can only take the values we list
    yes = "yes"
    no = "no"

my_answer = Answer.yes # from Answer, grab yes. dot accesses something belonging to the object or class named before it

print(my_answer)        # Answer.yes
print(my_answer.value)  # yes
```

# All together with LLMs

```python
# initialize the model
openai = OpenAI()

# Save your variables as a string
system_prompt = "You are a Talking Heads fan."
user_prompt = "Name a song by the Talking Heads."
PROMPT = [
    {"role": "user", "system": system_prompt, "content": user_prompt}
  ]

# model settings
# openai has a thing called chat, which has a thing called completions, which has a method called create. create takes arguements including...
client = openai.chat.completions.create(
    model = MODEL,
    messages = PROMPT,
    n = 3,
    temperature = 1
    )

client # this is an OpenAI object (created with a class). It contains lots of things we can access with dot notation
```

```python
client.choices
```

```python
client.choices[0]
```

```python
client.choices[0].message
```

```python
client.choices[0].message.content
```

```python
for response in client.choices:
  print(response.message.content)
```

```python
client = openai.chat.completions.create(
    model = "gpt-5.4",
    messages = PROMPT,
    n = 5,
    temperature = 2
    )

for response in client.choices:
  print(response.message.content)
```

```python
for row in df.index:
  print(row)
```

```python
responses = []

for row in df.index:

    song_title = df.loc[row, "song_title"]
    artist = df.loc[row, "artist"]

    user_prompt = f"""
    Song title: {song_title}
    Artist: {artist}

    Is this song by Talking Heads?
    Respond with only yes or no.
    """

    prompt = [
        {"role": "user", "content": user_prompt}
    ]

    completion = openai.chat.completions.create(
        model=MODEL,
        messages=prompt
    )

    response_text = completion.choices[0].message.content
    print(response_text)
    answer = Answer(response_text.strip().lower())

    responses.append(answer.value)


df["talking_heads"] = responses

df
```
