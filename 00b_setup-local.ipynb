## Setup to Run Locally
This set up assumes you're working from VS Code.

1. Confirm your working directory
2. Create a virtual environment + install dependencies
3. Save + load secrets
4. Run a test

---
### Confirm your working directory
When you opened VS Code to the llm-text-classification file, it should already set to llm-text-classification. To check, select Terminal > New Terminal from the dropdown on the top banner on VS Code. 

In the VS Code terminal, run `pwd` to print your working directory and ensure it ends in llm-text-classification. 

When you run `ls -a` you should see a file called .git, and another called requirements.txt.

### Create a virtual environment + install dependencies

Use the VS Code shortcut: CMD + SHIFT + P (on a Mac) or CTRL + SHIFT + P (on a PC) and select the following:
- Python: Create Environment 
- venv
- Python 3.12.3
- Name it `.venv` & press ENTER
- Select install project dependencies
- Check the box next to requirements.txt & click OK
- Select .venv as the kernel


### Save secrets + load secrets

- Create a copy of the file secrets-template.txt
- Rename the file to .env
- Open the file and add keys for `OPENAI_API_KEY` and `HF_TOKEN`
- Save the file
