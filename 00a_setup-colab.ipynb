## Setup to Run in Google Colab

This set up assumes you're working from the Google Colab UI.

1. Mount MyDrive
2. Set working directory
3. Install PyPi modules from requirements.txt
4. Save secrets
5. Load secrets to your environment
6. Run a test

```python
import os
from google.colab import drive
from google.colab import userdata
```

### Mount MyDrive

Mount your Google Drive so that your notebook in Colab has access to the files in `aime-con`.

```python
drive.mount('/content/drive/')
```

### Set Working Directory

Your working directory will need to point to the `aime-con` folder in your Google Drive. 

```python
# list the contents of MyDrive
# should see the aime-con folder
os.listdir("drive/MyDrive")
```

```python
os.chdir("./drive/MyDrive/aime-con")
```

```python
# this should match the contents listed in README.md
os.listdir(".")
```

### Install PyPi modules

All PyPi modules used are listed in `requirements.txt`. Use `-r` to install modules from this requirements file.

```python
!pip3 install -r requirements.txt
```

### Save secrets
Navigate to the left panel and click on 🔑 **Secrets**. Use **+ Add new secret** to add two new secrets called `OPENAI_API_KEY` and `HF_TOKEN`. Toggle Notebook access on for both.

<img src="https://storage.googleapis.com/generativeai-downloads/images/secrets.jpg" alt="You can find the Secrets tab on the left panel." width=50%>

### Load secrets

Load your secrets into you environment

```python
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
os.environ["HF_TOKEN"] = userdata.get('HF_TOKEN')
```

### Run a test
