# Making Non-Local Calls to LLMs
Non-local (API-based) calls are a way for developers to interact with large language models. You send a request over the internet to a provider's servers (e.g., OpenAI, Anthropic, Google) and receive a generated response back. 

Application programming interface (API) keys are how the provider verifies who is making the request (aka who to bill for it). That's where API keys come in; they're the credential that authenticates each request.

## Store API Keys
*Secrets management* is the process of controlling IT credentials, such as API keys, passwords, and configuration files. One approach for storing secrets is as environment variables. 

To do this: 
- Create a file called `.env` in your working directory 
- Ensure that `.env` is in your `.gitignore` file
- Add your keys to `.env`:
   ```
   OPENAI_API_KEY=sk-...
   ANTHROPIC_API_KEY=sk-ant-...
   ```
